1. Upload Data.

```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print(df)

```

```python
print(df.head(3))
print("Shape:", df.shape)
```

3. Display Dataset info.

```python
df.info()
```

4. Descrive data & Check missing values.

```python
print("descriptive statistics: ")
print(df.describe())

print("Missing Value: ")
print(df.isnull().sum())
```

5. Filter Dataset with Salary more than 40,000.

```python
print("Salary More than 40,000: ")
print(df[df["Salary"] > 40000])
```
