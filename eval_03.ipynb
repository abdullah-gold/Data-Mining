1. Create and Input dataset.

```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print(df)
```

2. Dispaly Dataset shape and information.

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

3. Identify and count missing values in each column using isnull().sum().

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

Replace missing values.

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Grade"] = df["Grade"].fillna(df["Grade"].mode()[0])

print("Dataset After Imputation:")
print(df)
```

Count duplicate records.

```python
df = df.drop(columns=["Unnamed: 0"], errors="ignore")
print("Number of Duplicate Records:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("After Removing Duplicates:")
print(df)
```

Final verification

```python
print("Remaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Records:")
print(df.duplicated().sum())

print("\nFinal Dataset Shape:")
print(df.shape)
```
